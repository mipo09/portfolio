# **Cafe Manager**

## **概要**

 これはカフェ経営シミュレーションゲームです。
 プレイヤーは自分のカフェに名前を付けて、仕入れや商品の価格変更などを行いながらカフェを経営します。  
 主なシステムとして営業開始するとランダムな数のお客さんが商品を注文し材料が足りていれば商品を提供して売り上げを得られます。また、営業中には「お祭り開催」「大雨」「テレビで紹介された」などのランダムイベントが一定の確率で発生し来客数や人気度に影響したり、修理費が発生します。営業を続けて一定の来客数を達成するとお店レベルが上がり、新しい材料とメニューが追加されます。  
 材料の仕入れでは日によって価格が変動する仕様、価格設定では商品の需要度が変化する仕様になっています。売上確認では売り上げの推移をグラフで表示したり総売り上げや総来客数、人気度などの経営データ、商品別の売り上げを確認できます。  
 所持金が25万円に到達するとゲームクリアとなり、反対に所持金が-1万円以下、もしくは人気度が-100になるとゲームオーバーとなります。

## **操作方法**

まず「ライブラリのインストール」というセルを実行してください。次に「Cafe Manager」というセルを実行すると遊べます。基本的にはボタンを押すことでゲームを進めることができます。  
タイトル画面の新しく遊ぶを押すと、カフェ名の入力画面に移動し、何も入力せずに「開店する」を押すと「My Cafe」という名前でゲームが開始します。  
ホーム画面では「営業開始」「仕入れ」「価格変更」「売上確認」「レシピ」「セーブしてタイトル」を選択できます。  

  <font color= green>営業開始</font> ：お客さんの数やイベントによって人気度、所持金が変化します。また、営業終了し「ホームへ戻る」を押すと次の日になります。  
  <font color= deepskyblue>仕入れ</font> ：それぞれの材料の右側にある＋ボタンと-ボタンで買う量を選択し、購入ボタンを押すと購入を確定することができます。  

  <font color= #D4AF37>価格変更</font> ：それぞれのメニューの価格を＋ボタンと-ボタンを使用して、10円単位で変更することができます。下の「適用」を選択することで価格変更を確定することができます。右側の星は需要度で価格が高くなるほど需要度が下がり、そのメニューを選ばれる確率が下がります。  

  <font color=#666666 >売上確認</font> ：上の「売上推移」「経営データ」「商品別売り上げ」を選択することでそれぞれを確認することができます。  

 <font color= saddlebrown>レシピ</font> ：現在作ることができるメニューの情報を確認することができます。  

<font color= orange>セーブして戻る</font>：その時点の所持金や人気度等の情報を保存し、タイトルに戻ります。その後に「続きから遊ぶ」を選択すると再開することができます。またタイトル画面の「セーブデータ削除」はその情報を削除します。  
※ランタイムをリセットすると、セーブデータが消える場合があります。

```python
#@title ===Cafe Manager===
import ipywidgets as widgets
from IPython.display import display,HTML, clear_output
import matplotlib.pyplot as plt
import japanize_matplotlib
import random
import time
import json
import os

#==初期値==
player={
    "cafe_name":"",
    "money":10000,
    "level":1,
    "total_sales":0,
    "total_customer":0,
    "total_left_customer":0,
    "next_level_customer": 50,
    "popularity":0,
    "day":1,
    "game_clear":False
}
#==在庫数==
inventory={
    "beans":0,
    "milk":0,
    "tea":0,
    "sugar":0,
    "flour": 0,
    "egg": 0,
    "cream": 0,
    "chocolate":0,
    "strawberry":0,
}
#材料データ
ingredient_data={
    "beans":{
        "name":"コーヒー豆",
        "base_price":50,
        "price":50,
        "level":1
    },
    "milk":{
        "name":"牛乳",
        "base_price":80,
        "price":80,
        "level":2
    },
    "tea":{
        "name":"茶葉",
        "base_price":60,
        "price":60,
        "level":2
    },
    "sugar":{
        "name":"砂糖",
        "base_price":30,
        "price":30,
        "level":2
    },
    "flour": {
        "name":"小麦粉",
        "base_price":40,
        "price":40,
        "level":3
    },
    "egg": {
        "name":"卵",
        "base_price":35,
        "price":35,
        "level":3
    },
    "cream": {
        "name":"生クリーム",
        "base_price":120,
        "price":120,
        "level":7
    },
    "chocolate":{
        "name":"チョコレート",
        "base_price":100,
        "price":100,
        "level":3
    },
    "strawberry":{
        "name":"イチゴ",
        "base_price":150,
        "price":150,
        "level":7
    }
}

#メニュー価格
menu_data = {
    "Coffee": {
        "name": "コーヒー",
        "base_price": 300,
        "price": 300,
        "ingredients": {
            "beans": 1
        }
    },
    "Milk Tea": {
        "name": "ミルクティー",
        "base_price": 400,
        "price": 400,
        "ingredients": {
            "tea": 1,
            "milk": 1,
            "sugar": 1
        }
    },
    "Cookie": {
        "name": "クッキー",
        "base_price": 300,
        "price": 300,
        "ingredients": {
            "flour": 1,
            "egg":1,
            "sugar": 1,
            "chocolate": 1
        }
    },
    "Cafe Latte": {
        "name": "カフェラテ",
        "base_price": 500,
        "price": 500,
        "ingredients": {
            "beans": 1,
            "milk": 2
        }
    },
    "Shortcake": {
        "name": "ショートケーキ",
        "base_price": 500,
        "price": 500,
        "ingredients": {
            "flour": 1,
            "egg": 1,
            "cream": 1,
            "strawberry": 2
        }
    },
    "Strawberry Chocolate Parfait": {
        "name": "イチゴチョコパフェ",
        "base_price": 700,
        "price": 700,
        "ingredients": {
            "flour":1,
            "egg":1,
            "cream": 1,
            "chocolate": 2,
            "strawberry": 2
        }
    }
}

unlock_menu = {
    2: ["Milk Tea"],
    3: ["Cookie"],
    5: ["Cafe Latte"],
    7: ["Shortcake"],
    10: ["Strawberry Chocolate Parfait"]
}

customer_menu = ["Coffee"]

sales_history = []
player["total_sales"] = 0
for item in menu_data:
    menu_data[item]["total_sales"] = 0

#タイトル画面
def show_title():
    clear_output()
    # タイトル
    title = HTML("""
    <h1 style="text-align:center; color:brown;">
        ☕ Cafe Manager ☕
    </h1>
    <h3 style="text-align:center;">
        あなたのカフェを経営しよう！
    </h3>
    """)

    display(title)

    # ボタン
    new_button = widgets.Button(
        description="▶ 新しく始める",
        button_style="success",
        layout=widgets.Layout(width="300px", height="45px")
    )

    load_button = widgets.Button(
        description="📂 続きから遊ぶ",
        button_style="info",
        layout=widgets.Layout(width="300px", height="45px")
    )

    delete_button = widgets.Button(
        description="🗑 セーブデータ削除",
        button_style="danger",
        layout=widgets.Layout(width="300px", height="45px")
    )

   #それぞれのボタン
    def new_game(b):
        clear_output()
        reset_game()
        show_name_input()

    def load_save_button(b):
        clear_output()
        load_game()

    def delete_save_button(b):
        clear_output()
        delete_save()

    new_button.on_click(new_game)
    load_button.on_click(load_save_button)
    delete_button.on_click(delete_save_button)

    menu = widgets.VBox(
        [new_button, load_button, delete_button],
        layout=widgets.Layout(
            align_items="center",
            gap="10px"
        )
    )

    display(menu)

#初期化
def reset_game():
    global player
    global inventory
    global menu_data
    global ingredient_data
    global customer_menu
    global sales_history
    player = {
        "cafe_name": "",
        "money": 10000,
        "level": 1,
        "total_sales": 0,
        "total_customer": 0,
        "total_left_customer": 0,
        "next_level_customer": 50,
        "popularity": 0,
        "day": 1,
        "game_clear":False
    }
    inventory = {
        "beans": 0,
        "milk": 0,
        "tea": 0,
        "sugar": 0,
        "flour": 0,
        "egg": 0,
        "cream": 0,
        "chocolate": 0,
        "strawberry": 0
    }

    for item in menu_data:
        menu_data[item]["price"] = menu_data[item]["base_price"]
        menu_data[item]["total_sales"] = 0
    for item in ingredient_data:
        ingredient_data[item]["price"] = ingredient_data[item]["base_price"]
    customer_menu = ["Coffee"]
    sales_history = []

#名前入力画面
def show_name_input():

    clear_output()

    display(HTML("""
    <h2 style="text-align:center;">
    カフェの名前を入力してください
    </h2>
    """))

    cafe_name = widgets.Text(
        placeholder="例：○○ Cafe",
        layout=widgets.Layout(width="300px")
    )

    decide_button = widgets.Button(
        description="開店する",
        button_style="success"
    )
    display(
        widgets.VBox(
            [cafe_name, decide_button],
            layout=widgets.Layout(
                width="100%",
                align_items="center",
                gap="15px"
            )
        )
    )
    def start(b):
      name = cafe_name.value.strip()
      if name == "":
        name = "My Cafe"
      player["cafe_name"]=name
      clear_output()
      show_main()

    update_market()
    decide_button.on_click(start)

#==ホーム画面==
def show_main():
  clear_output()
  display(HTML(f"""
    <h1 style="text-align:center; color:brown;">
        ☕ {player["cafe_name"]} ☕
    </h1>

    <h3 style="text-align:center;">
        Day {player["day"]}
    </h3>
    """))
  # ステータス表示
  display(HTML(f"""
    <div style="text-align:center;">
    <table style="margin:auto; font-size:20px;">
      <tr>
        <td style="padding:0 30px;">💰 所持金：{player["money"]} 円</td>
        <td style="padding:0 30px;">⭐ 人気度：{player["popularity"]}</td>
        <td style="padding:0 30px;">🏪 レベル：Lv.{player["level"]}</td>
      </tr>
    </table>
    </div>
    """))
  display(HTML("<h3 style='text-align:center;'>📦 在庫</h3>"))
  stock=""
  for key, data in ingredient_data.items():
    if player["level"]>=data["level"]:
      stock+=(f"{data['name']}:{inventory[key]}個　")
  display(HTML(f"""
  <div style="
  text-align:center;
  font-size:18px;
  line-height:2;
  ">
  {stock}
  </div>
  """))

   # meinボタン
  open_button = widgets.Button(
      description="営業開始",
      button_style="success",
      layout=widgets.Layout(width="160px", height="45px")
  )
  shop_button = widgets.Button(
      description="仕入れ",
      button_style="info",
      layout=widgets.Layout(width="160px", height="45px")
  )
  price_button = widgets.Button(
      description="価格変更",
      layout=widgets.Layout(width="160px", height="45px")
  )
  price_button.style.button_color = "#FFD700"
  sales_button = widgets.Button(
      description="売上確認",
      layout=widgets.Layout(width="160px", height="45px")
  )
  recipe_button =widgets.Button(
      description="レシピ",
      layout=widgets.Layout(width="160px",height="45px")
  )
  recipe_button.style.button_color = "#D2B48C"
  save_button = widgets.Button(
      description="セーブしてタイトル",
      button_style="warning",
      layout=widgets.Layout(width="160px", height="45px")
  )
  #ボタン並べる
  display(
      widgets.VBox([
        widgets.HBox(
            [open_button,shop_button,price_button],
            layout=widgets.Layout(
                justify_content="center",
                gap="10px"
            )
         ),
        widgets.HBox(
            [sales_button,recipe_button,save_button],
            layout=widgets.Layout(
                justify_content="center",
                gap="10px",
                margin="10px 0 0 0"
            )
        )
    ])
  )
  def open_shop_button(b):
    clear_output()
    open_shop()
  def shop(b):
    clear_output()
    show_market()
  def price(b):
    clear_output()
    show_price_menu()
  def sales(b):
    clear_output()
    show_sales()
  def recipe(b):
    clear_output()
    show_recipe()
  def save(b):
    clear_output()
    save_game()
    display(HTML("""
        <h3 style="text-align:center;color:green;">
        ✔ セーブしました！
        </h3>
        """))
    time.sleep(1)
    show_title()

  open_button.on_click(open_shop_button)
  shop_button.on_click(shop)
  price_button.on_click(price)
  sales_button.on_click(sales)
  recipe_button.on_click(recipe)
  save_button.on_click(save)


#==営業開始==
def open_shop():
  #イベント
  event = event_system()
  player["popularity"] += event["popularity"]
  # 修理費
  if event["repair_cost"] > 0:
      player["money"] -= event["repair_cost"]


  #選ばれやすさ
  weights = []
  for item in customer_menu:
    base = menu_data[item]["base_price"]
    price = menu_data[item]["price"]
    diff = price-base
    weight=max(10,100-diff)
    weights.append(weight)

  #客数
  customers = int(random.randint(5, 15) * max(0.5,1+ player["popularity"] /100)* event["customer_rate"])

  order_count = {item: 0 for item in customer_menu}
  left_customer=0
  miss_customer=0
  sales=0

  for _ in range(customers):
    sold = 0
    miss =0
    orders = random.choices(
      customer_menu,
      weights=weights,
      k=random.randint(1, 2)
    )
    for order in orders:
      can_make = True

      for ingredient,amount in menu_data[order]["ingredients"].items():
        if inventory[ingredient]<amount:
          can_make= False
          break
      if can_make:
        for ingredient,amount in menu_data[order]["ingredients"].items():
          inventory[ingredient]-= amount
        order_count[order] +=1
        sales += menu_data[order]["price"]
        sold +=1
      else:
        #50%の確率で別のメニューを注文
        if random.random()<0.5:
          available = []
          for item in customer_menu:
            ok = True
            for ingredient,amount in menu_data[item]["ingredients"].items():
              if inventory[ingredient]<amount:
                ok = False
                break
            if ok:
              available.append(item)
          if available:
            new_order = random.choice(available)
            for ingredient, amount in menu_data[new_order]["ingredients"].items():
              inventory[ingredient] -= amount
            order_count[new_order]+=1
            sales += menu_data[new_order]["price"]
            sold +=1
          else:
            miss+=1
        else:
          miss+=1
    if sold == 0:
      left_customer += 1
    elif miss>0:
      miss_customer+=1

   #データ更新
  popularity_change = (customers-left_customer-miss_customer)//2 - left_customer*2 - miss_customer
  player["total_customer"]+=customers-left_customer
  player["total_left_customer"]+=left_customer
  player["money"]+=sales
  player["popularity"]+=popularity_change
  player["popularity"]=max(-100,min(100,player["popularity"]))
  # == ゲームオーバーチェック ==
  if player["money"] <= -10000:
    show_game_over("所持金が -10,000円以下になったため、閉店しました。")
    return
  if player["popularity"] <= -100:
    show_game_over("人気度が -100になったため、閉店しました。")
    return

  player["total_sales"] += sales

  if event["repair_cost"] > 0:
    repair_text = f"🔧 修理費：-{event['repair_cost']}円<br><br>"
  else:
    repair_text = ""

  sales_history.append(sales)
  for item,count in order_count.items():
    if "total_sales" not in menu_data[item]:
      menu_data[item]["total_sales"] = 0
    menu_data[item]["total_sales"]+=count

  result =""
  for item,count in order_count.items():
    if count>0:
      result += f'{menu_data[item]["name"]}:{count}個<br>'

  for i in range(6):
    clear_output()
    dots ="."*i
    display(HTML(f"""
    <h1 style="text-align:center; color:brown;">
     ☕ 営業中{dots}
    </h1>
    """))
    time.sleep(1)
  clear_output()

  #レベルチェック
  level_result=""
  if player["next_level_customer"]<500:
    while player["total_customer"] >= player["next_level_customer"]:
      player["level"] += 1
      player["next_level_customer"] += 50
      level_result += f"🎉 お店レベルが Lv.{player['level']} に上がりました！<br>"


  #追加メニューチェック
  menu_result=""
  for level, menu in unlock_menu.items():
    if player["level"] >= level:
      for item in menu:
          if item not in customer_menu:
              customer_menu.append(item)
              menu_result += f"🆕 {menu_data[item]['name']} が追加されました！<br>"


  display(HTML(f"""
  <h1 style="text-align:center; color:green;">
  ☕営業終了
  </h1>

  <h3 style="text-align:center;">
  🎊 今日のイベント：{event["title"]}<br><br>
  {repair_text}
  💰 売上：{sales} 円<br><br>

  👥 来客数：{customers} 人<br><br>

  🚶 帰ったお客さん：{left_customer} 人<br><br>

  ⭐ 人気度：{player["popularity"]}({popularity_change+event["popularity"]:+})<br><br>
  🍽売れた商品<br><br>
  {result}<br>
  {level_result}
  {menu_result}
  </h3>
  """))





  #ホームへ戻るボタン
  home_button = widgets.Button(
      description="ホームへ戻る",
      button_style="info",
      layout=widgets.Layout(width="200px",height="45px")
  )
  display(
      widgets.HBox(
        [home_button],
        layout=widgets.Layout(
            justify_content="center"
            )
      )
  )
  def back_home(b):
    # == クリア判定 ==
    if player["money"] >= 250000 and not player["game_clear"]:
      player["game_clear"] = True
      show_clear()
      return

    for i in range(4):
      clear_output()
      dots ="."*i
      display(HTML(f"""
      <h1 style="text-align:center; color:navy;">
       🌃次の日{dots}
      </h1>
      """))
      time.sleep(1)
    player["day"]+=1
    update_market()
    clear_output()
    show_main()
  home_button.on_click(back_home)

#ショップ価格の更新
def update_market():
  for item in ingredient_data.values():
    rate = random.uniform(0.8,1.2)
    item["price"]=int(item["base_price"]*rate)

#==仕入れ画面==
def show_market():
  clear_output()
  display(HTML("""
    <h2 style="text-align:center;">
    🛒 仕入れ
    </h2>
    """))

  display(HTML(f"""
    <h3 style="text-align:center;">
    💰 所持金：{player["money"]} 円
    </h3>
    """))

  # 購入数
  buy_count = {}

  # 合計金額表示
  total_label = widgets.HTML(
        value="<h3 style='text-align:center;'>合計金額：0円</h3>"
  )

  # 合計金額更新
  def update_total():

    total = 0

    for key in buy_count:
      total += buy_count[key]["count"] * ingredient_data[key]["price"]

    total_label.value = f"<h3 style='text-align:center;'>合計金額：{total}円</h3>"

  # 材料表示
  for key, data in ingredient_data.items():
    if player["level"] < data["level"]:
        continue
    count = {"count": 0}
    name = widgets.Label(
        value=data["name"],
        layout=widgets.Layout(width="120px")
    )
    price = widgets.Label(
        value=f'{data["price"]}円',
        layout=widgets.Layout(width="70px")
    )
    stock = widgets.Label(
        value=f'在庫 {inventory[key]}',
        layout=widgets.Layout(width="80px")
    )
    number = widgets.Label(
        value="0",
        layout=widgets.Layout(width="30px")
    )
    plus = widgets.Button(
        description="+",
        layout=widgets.Layout(width="40px")
    )
    minus = widgets.Button(
        description="-",
        layout=widgets.Layout(width="40px")
    )
    buy_count[key] = count
    # ＋ボタン
    def plus_click(b, c=count, label=number):
        c["count"] += 1
        label.value = str(c["count"])
        update_total()
    # －ボタン
    def minus_click(b, c=count, label=number):
        if c["count"] > 0:
          c["count"] -= 1
          label.value = str(c["count"])
          update_total()
    plus.on_click(plus_click)
    minus.on_click(minus_click)
    display(
        widgets.HBox(
            [name, price,stock,minus,number,plus],
            layout=widgets.Layout(
                justify_content="center",
                align_items="center",
                margin="5px 0"
            )
        )
    )

  display(total_label)
  purchase_button = widgets.Button(
        description="購入する",
        button_style="success",
        layout=widgets.Layout(width="150px")
  )

  home_button = widgets.Button(
    description="ホームへ戻る",
    button_style="info",
    layout=widgets.Layout(width="150px")
  )

  # 購入処理
  def purchase(b):
    total = 0
    for key in buy_count:
      total += buy_count[key]["count"] * ingredient_data[key]["price"]
    if total > player["money"]:
      total_label.value = (
        "<h3 style='color:red;text-align:center;'>"
        "所持金が足りません！"
        "</h3>"
      )
      return
    player["money"] -= total
    for key in buy_count:
      inventory[key] += buy_count[key]["count"]
    show_market()
  purchase_button.on_click(purchase)
  home_button.on_click(lambda b: show_main())
  display(
        widgets.HBox(
            [purchase_button, home_button],
            layout=widgets.Layout(
                justify_content="center",
                gap="20px"
            )
        )
    )



#＝＝価格変更画面＝＝
def show_price_menu():
    clear_output()
    display(HTML("""
    <h2 style="text-align:center;">
    💰 価格変更
    </h2>
    """))
    display(HTML(f"""
    <h3 style="text-align:center;">
    ☕ {player["cafe_name"]}
    </h3>
    """))
    price_data={}
    #需要
    def update_star(item,label):
      base=menu_data[item]["base_price"]
      price=price_data[item]["price"]
      diff=price-base
      weight=max(10,100-diff)

      if weight>=90:
        star = "★★★★★"
      elif weight >= 70:
        star = "★★★★☆"
      elif weight >= 50:
        star = "★★★☆☆"
      elif weight >= 30:
        star = "★★☆☆☆"
      else:
        star = "★☆☆☆☆"
      label.value=f"需要{star}"

    for item in customer_menu:
        data = menu_data[item]
        price_data[item] = {
            "price": data["price"]
        }
        name = widgets.Label(
            value=data["name"],
            layout=widgets.Layout(width="160px")
        )
        price_label = widgets.Label(
            value=f'{data["price"]}円',
            layout=widgets.Layout(width="70px")
        )
        star_label = widgets.Label(
            layout=widgets.Layout(width="120px")
        )
        update_star(item, star_label)
        plus = widgets.Button(
            description="+",
            layout=widgets.Layout(width="40px")
        )
        minus = widgets.Button(
            description="-",
            layout=widgets.Layout(width="40px")
        )
        # ＋
        def plus_click(b,
                       item=item,
                       label=price_label,
                       star=star_label):

            price_data[item]["price"] += 10

            label.value = f'{price_data[item]["price"]}円'

            update_star(item, star)
         # －
        def minus_click(b,item=item,label=price_label,star=star_label):
            if price_data[item]["price"] > 10:
                price_data[item]["price"] -= 10
                label.value = f'{price_data[item]["price"]}円'
                update_star(item, star)
        plus.on_click(plus_click)
        minus.on_click(minus_click)
        display(
            widgets.HBox(
                [name,minus,price_label,plus,star_label],
                layout=widgets.Layout(
                    width="100%",
                    justify_content="center",
                    align_items="center",
                    gap="10px"
                )
            )
        )

    display(HTML("<br>"))

    apply_button = widgets.Button(
        description="適用",
        button_style="success",
        layout=widgets.Layout(width="150px")
    )

    home_button = widgets.Button(
        description="ホームへ戻る",
        button_style="info",
        layout=widgets.Layout(width="150px")
    )
    def apply_price(b):
        for item in customer_menu:
            menu_data[item]["price"] = price_data[item]["price"]
        clear_output()
        display(HTML("""
        <h3 style="text-align:center;color:green;">
        ✔ 価格を変更しました！
        </h3>
        """))
        time.sleep(1)
        show_main()

    def back_home(b):
        show_main()
    apply_button.on_click(apply_price)
    home_button.on_click(back_home)
    display(
        widgets.HBox(
            [apply_button, home_button],
            layout=widgets.Layout(
                width="100%",
                justify_content="center",
                gap="20px"
            )
        )
    )

#==売上確認画面==
def show_sales():
  clear_output()
  display(HTML("""
  <h2 style="text-align:center;">
    📊 売上確認
  </h2>
  """))
  output = widgets.Output()
  #売上推移
  def show_graph(b=None):
    with output:
      clear_output(wait=True)
      plt.figure(figsize=(8,4))
      plt.plot(
          range(1,len(sales_history)+1),
          sales_history,
          marker="o"
      )
      plt.title("売上推移")
      plt.xlabel("Day")
      plt.ylabel("売上(円)")
      plt.xticks(range(1, len(sales_history)+1))
      plt.grid()
      plt.tight_layout()
      plt.show()
  #経営データ
  def show_status(b=None):
    with output:
      clear_output(wait=True)
      display(HTML(f"""
      <div style="text-align:center;">
      <h3>経営データ</h3>
      <b>💰 所持金</b>：{player["money"]} 円<br><br>
      <b>💵 総売上</b>：{player["total_sales"]} 円<br><br>
      <b>👥 総来客数</b>：{player["total_customer"]} 人<br><br>
      <b>🚶 帰った人数</b>：{player["total_left_customer"]} 人<br><br>
      <b>⭐ 人気度</b>：{player["popularity"]}<br><br>
      <b>🏪 レベル</b>：Lv.{player["level"]}<br><br>
      <b>📅 営業日数</b>：{player["day"]} 日
      </div>
      """))
  #商品別売り上げ
  def show_menu_sales(b=None):
    with output:
      clear_output(wait=True)
      names=[]
      values=[]
      for item in customer_menu:
        names.append(menu_data[item]["name"])
        values.append(menu_data[item]["total_sales"])
      plt.figure(figsize=(8,4))
      plt.bar(names,values)
      plt.title("商品別売り上げ")
      plt.ylabel("販売数")
      plt.tight_layout()
      plt.show()
  graph_button=widgets.Button(
      description="売上推移",
      button_style="info",
      layout=widgets.Layout(width="150px")
  )
  status_button=widgets.Button(
      description="経営データ",
      button_style="warning",
      layout=widgets.Layout(width="150px")
  )
  menu_button=widgets.Button(
      description="商品別売り上げ",
      button_style="success",
      layout=widgets.Layout(width="150px")
  )
  home_button=widgets.Button(
      description="ホームへ戻る",
      button_style="primary",
      layout=widgets.Layout(width="180px")
  )
  graph_button.on_click(show_graph)
  status_button.on_click(show_status)
  menu_button.on_click(show_menu_sales)
  home_button.on_click(lambda b: show_main())

  display(
      widgets.HBox([graph_button,status_button,menu_button],
      layout=widgets.Layout(
          justify_content="center",
          gap="20px"
      )
    )
  )
  display(output)
  display(
      widgets.HBox([home_button],
      layout=widgets.Layout(
          justify_content="center",
          margin="20px 0"
      )
    )
  )
  show_status()

#==イベント==
def event_system():
  event = {
        "title": "通常営業",
        "customer_rate": 1.0,
        "popularity": 0,
        "repair_cost": 0
    }
  r = random.randint(1,100)
  if r <= 5:
    event["title"] = "🎉 お祭り開催！"
    event["customer_rate"] = 2.0
  elif r <= 15:
    event["title"] = "☔ 大雨..."
    event["customer_rate"] = 0.6
  elif r <= 18:
    event["title"] = "📺 テレビで紹介された！"
    event["customer_rate"] = 1.5
    event["popularity"] = 15
  elif r <= 22:
    event["title"] = "⭐ SNSで話題！"
    event["popularity"] = 10
  elif r <= 25:
    event["title"] = "😡 クレーム発生..."
    event["popularity"] = -20
  # 新しく追加
  elif r <= 30:
    cost = random.randint(500*player["level"], 3000*player["level"])
    event["title"] = f"🔧 コーヒーマシンが故障！ 修理費 {cost}円"
    event["repair_cost"] = cost
  return event

def show_recipe():
  clear_output()
  display(HTML("""
    <h2 style="text-align:center;">
        📖 レシピ
    </h2>
    """))
  for item in customer_menu:
    data = menu_data[item]

    # 材料表示
    ingredients = ""

    for ingredient, amount in data["ingredients"].items():
      ingredient_name = ingredient_data[ingredient]["name"]
      ingredients += f"{ingredient_name} × {amount}<br>"

    display(HTML(f"""
      <div style="
        width:400px;
        margin:10px auto;
        padding:15px;
        border:2px solid #ccc;
        border-radius:10px;
        text-align:center;
      ">
        <h3>☕ {data["name"]}</h3>

        <p>
            <b>販売価格：</b>{data["price"]} 円
        </p>

        <p>
            <b>必要な材料</b><br>
            {ingredients}
        </p>
    </div>
    """))

  home_button=widgets.Button(
      description="ホームへ戻る",
      button_style="info",
      layout=widgets.Layout(
          width="180px",
          height="45px"
        )
    )
  home_button.on_click(lambda b:show_main())
  display(
        widgets.HBox(
            [home_button],
            layout=widgets.Layout(
                justify_content="center",
                margin="20px 0"
            )
        )
    )

#==セーブ==
def save_game():
    save_data = {
        "player": player,
        "inventory": inventory,
        "menu_data": menu_data,
        "ingredient_data": ingredient_data,
        "customer_menu": customer_menu,
        "sales_history": sales_history
    }
    with open("save_data.json", "w", encoding="utf-8") as f:
        json.dump(save_data, f, ensure_ascii=False, indent=4)

#==ロード==
def load_game():
    global player
    global inventory
    global menu_data
    global ingredient_data
    global customer_menu
    global sales_history
    try:
        with open("save_data.json", "r", encoding="utf-8") as f:
            save_data = json.load(f)
        player = save_data["player"]
        inventory = save_data["inventory"]
        menu_data = save_data["menu_data"]
        ingredient_data = save_data["ingredient_data"]
        customer_menu = save_data["customer_menu"]
        sales_history = save_data["sales_history"]
        clear_output()
        display(HTML("""
        <h3 style="text-align:center;color:green;">
        ✔ セーブデータをロードしました！
        </h3>
        """))
        time.sleep(1)
        clear_output()
        show_main()
    except FileNotFoundError:
      clear_output()
      display(HTML("""
        <h3 style="text-align:center;color:red;">
        ✘ セーブデータがありません。
        </h3>
        """))
      time.sleep(1)
      show_title()

#==セーブデータ消去==
def delete_save():
    if os.path.exists("save_data.json"):
      os.remove("save_data.json")
      clear_output()
      reset_game()
      display(HTML("""
      <h3 style="text-align:center;color:red;">
      ✘ セーブデータを消去しました。
      </h3>
      """))
      time.sleep(1)
      show_title()
    else:
      clear_output()
      display(HTML("""
      <h3 style="text-align:center;color:red;">
      ✘ セーブデータがありません。
      </h3>
      """))
      time.sleep(1)
      show_title()


#==ゲームクリア画面==
def show_clear():
    clear_output()
    display(HTML("""
    <h1 style="text-align:center; color:gold;">
        🎉 おめでとうございます！ 🎉
    </h1>
    <h2 style="text-align:center; color:brown;">
        カフェの所持金が25万円に到達しました！
    </h2>
    <h3 style="text-align:center;">
        あなたのカフェは大成功です！
    </h3>
    <br>
    """))
    continue_button = widgets.Button(
        description="▶ ゲームを続ける",
        button_style="success",
        layout=widgets.Layout(
            width="200px",
            height="45px"
        )
    )
    title_button = widgets.Button(
        description="🏠 タイトルに戻る",
        button_style="info",
        layout=widgets.Layout(
            width="200px",
            height="45px"
        )
    )

    def continue_game(b):
        clear_output()
        show_main()
    def return_title(b):
        clear_output()
        show_title()

    continue_button.on_click(continue_game)
    title_button.on_click(return_title)

    display(
        widgets.HBox(
            [continue_button, title_button],
            layout=widgets.Layout(
                justify_content="center",
                gap="20px"
            )
        )
    )

#==ゲームオーバー画面==
def show_game_over(reason):
    clear_output()

    display(HTML(f"""
    <h1 style="text-align:center; color:red;">
        ☕ 閉店しました
    </h1>
    <h2 style="text-align:center;">
        GAME OVER
    </h2>
    <p style="text-align:center; font-size:20px;">
        {reason}
    </p>
    <hr>
    <div style="text-align:center; font-size:18px;">
        <p>💰 所持金：{player["money"]} 円</p>
        <p>⭐ 人気度：{player["popularity"]}</p>
        <p>👥 総来客数：{player["total_customer"]} 人</p>
        <p>💵 総売上：{player["total_sales"]} 円</p>
        <p>📅 営業日数：{player["day"]} 日</p>
    </div>
    """))
    title_button = widgets.Button(
        description="タイトルへ戻る",
        button_style="danger",
        layout=widgets.Layout(
            width="200px",
            height="45px"
        )
    )
    display(
        widgets.HBox(
            [title_button],
            layout=widgets.Layout(
                justify_content="center",
                margin="20px 0"
            )
        )
    )

    title_button.on_click(lambda b: show_title())



#==実行==
show_title()
```
